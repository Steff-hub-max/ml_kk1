**2. Förklara hur OvR- och OvO-algoritmerna fungerar.**

Många enkla datormodeller kan bara välja mellan två alternativ (ja/nej). För att de ska kunna välja mellan flera kategorier (t.ex. Hund, Katt, Häst) använder man två olika knep:

OvR (One-vs-Rest):
Datorn skapar en modell för varje djur.

Modell 1 testar: "Är det en Hund eller NÅGOT ANNAT?"

Modell 2 testar: "Är det en Katt eller NÅGOT ANNAT?"

Modell 3 testar: "Är det en Häst eller NÅGOT ANNAT?"
Den modell som känns säkrast på sin sak vinner.

OvO (One-vs-One):
Datorn ställer alla kategorier mot varandra två och två i en turnering.

Match 1: Hund vs Katt

Match 2: Hund vs Häst

Match 3: Katt vs Häst
Kategorin som vinner flest matcher vinner totalt.

**3. Förklara följande utvärderingsmått:**

a) Confusion matrix (Förväxlingsmatris):
En enkel tabell som visar hela sanningen. Den visar hur många gånger datorn gissade rätt och exakt vilka saker den förväxlade med varandra.

b) Accuracy (Total pricksäkerhet):
Hur stor andel av ALLA gissningar som var rätt.

Exempel: Om datorn gissar rätt på 90 av 100 mejl är pricksäkerheten 90 procent.

c) Precision (Träffsäkerhet):
När datorn väl säger "Det här är skräppost", hur ofta stämmer det?

Högt värde betyder: Datorn ropar sällan "varg" i onödan.

d) Recall (Fångstförmåga):
Hur stor andel av ALL den riktiga skräpposten lyckades datorn faktiskt hitta?

Högt värde betyder: Datorn missar nästan inga skräpmejl, men kan råka ta med några vanliga mejl av misstag.

e) F1-score (Balansmåttet):
Ett samlat betyg som slår ihop Precision och Recall till ett enda tal. Det visar hur väl datorn balanserar båda kraven.

f) ROC-kurvan:
Ett diagram som visar hur bra datorn är på att göra rätt jämfört med hur ofta den gör fel, beroende på hur sträng eller förlåtande man gör den.

**4. Vad är precision-recall tradeoff för något?**

Det handlar om en inbyggd kompromiss (en dragkamp). Du kan nästan aldrig få 100 procent på både Precision och Recall samtidigt.

Om du gör datorn väldigt sträng (Hög Precision):
Den stämplar bara ett mejl som skräppost om den är helt säker. Du slipper få viktiga mejl kastade i skräpkorgen, men en del skräppost kommer att slinka igenom till din inkorg (Låg Recall).

Om du gör datorn väldigt snäll (Hög Recall):
Den fångar upp all skräppost som finns, men den blir för ivrig och börjar kasta bort dina vanliga, viktiga mejl också (Låg Precision).

Du måste därför välja vad som är viktigast för just ditt problem.

**5. Några vanligt förekommande modeller för klassificeringsproblem är enligt
nedan. Förklara översiktligt hur respektive modell fungerar. Läs även ige-
nom respektive modells dokumentation, notera att du inte behöver förstå
alla detaljer från dokumentationen men det är bra att ha läst igenom den**

**a) Logistisk regression (LogisticRegression)**
Hur den fungerar: Trots namnet är detta en klassificeringsmodell. Den beräknar en linjär kombination av indata (features) och kör resultatet genom en s-formad funktion (sigmoid-funktionen) för att omvandla värdet till en sannolikhet mellan 0 och 1. Om sannolikheten överstiger en visströskel (oftast 0,5) klassificeras datapunkten som klass 1, annars klass 0.

**b) Support Vector Machines (SVC)**
Hur den fungerar: SVM försöker hitta en linje eller ett plan (hyperplan) som separerar klasserna med så stor marginal som möjligt till de närmaste datapunkterna (som kallas support vectors). Med hjälp av "kernel-tricket" kan modellen även hantera svåra, icke-linjära gränser genom att lyfta datan till högre dimensioner.

**c) Beslutsträd (DecisionTreeClassifier)**
Hur den fungerar: Skapar en trädliknande struktur av ja/nej-frågor baserat på indata (t.ex. "Är carat > 0.5?"). För varje fråga delas datan upp i allt renare delmängder tills den når ett slutgiltigt löv som ger klassen. Det är en mycket lättförståelig och visuell modell.

**d) Ensemble learning (VotingClassifier & BaggingClassifier)**
Hur det fungerar: Ensemble learning kombinerar flera enskilda modeller för att få en bättre och mer stabil förutsägelse än vad enstaka modeller klarar på egen hand.

VotingClassifier: Tränar helt olika modeller (t.ex. Logistisk regression, SVM och Beslutsträd) och låter dem rösta om den slutgiltiga klassen (majoritetsröstning).

BaggingClassifier: Tränar samma typ av modell flera gånger på slumpmässiga delmängder av träningsdatan (med återläggning) och tar genomsnittet eller röstningen av dem.

**e) Random Forest (RandomForestClassifier)**
Hur den fungerar: En vidareutveckling av Bagging med beslutsträd. Den bygger en hel "skog" av många oberoende beslutsträd. Varje träd tränas på ett slumpmässigt urval av datan och får bara välja bland ett slumpmässigt urval av attribut vid varje split. Slutgiltigt resultat blir majoritetsrösten från alla träd, vilket kraftigt minskar risken för överanpassning.

**f) Extra Trees (ExtraTreesClassifier)**
Hur den fungerar: Står för Extremely Randomized Trees. Fungerar väldigt likt Random Forest, men tar slumpmässigheten ett steg längre: istället för att leta efter den optimala tröskeln för att dela upp datan vid varje nod, väljer den uppdelningströsklar helt slumpmässigt. Detta gör träningen extremt snabb och kan ge ännu lägre varians.

**7. Stina säger till Kalle på lunchsamtalet “jag vill ha högsta möjliga precision
för vår klassificieringsmodell”. Kalle funderar ett tag och säger “men vad
händer då med recall”? Vad hade du svarat? I vilka fall kan man tänka sig
vilja ha en så hög precision som möjligt? I vilka fall kan det vara dåligt?
Om vi tänker oss rättsväsendet där en slutgiltig dom kan leda till fängelse,
vad kan vi då säga om precision-recall tradeoff?**

* **Svar till Kalle & Stina:**  
  Om vi tvingar modellen att ha så **hög precision som möjligt** blir den extremt försiktig och gissar bara när den är 100 % säker. Det gör att **recall sjunker** drastiskt – modellen blir så feg att den missar massor av riktiga fall.

* **När vill vi ha så hög precision som möjligt?**  
  När kostnaden för ett **falskt larm (False Positive)** är väldigt hög.  
  * *Exempel:* Ett skräppostfilter för e-post. Du vill inte att vanliga, viktiga mejl råkar hamnar i skräpkorgen.

* **När kan hög precision vara dåligt?**  
  När det är farligt eller dyrt att **missa ett riktigt fall (False Negative)**.  
  * *Exempel:* Cancerdiagnostik eller brandlarm. Det är bättre med ett falskt larm för mycket än att missa en sjuk patient eller en riktig brand.

* **Tillämpning i rättsväsendet (fängelsestraff):**  
  Här tillämpar man principen *"hellre fria än fälla"*. Det kräver **extremt hög precision** – vi vill inte sätta en oskyldig person i fängelse. Baksidan blir lägre recall: en del faktiska brottslingar kommer att gå fria för att bevisen inte räckte hela vägen upp till den höga kraven.

### 9: Logiken bakom `.fit_transform()` vs `.transform()`

Principen handlar om att förhindra **dataläckage (data leakage)** så att modellen inte råkar "tjuvtitta" på testdatan i förväg.

Tänk dig att du ska räkna ut medelvikten på en skolklass:

1. **`.fit()` = Att lära sig reglerna.** Du väger alla i träningsklassen och räknar ut att medelvikten är **60 kg**.
2. **`.transform()` = Att använda reglerna.** Du drar av 60 kg från varje elevs vikt så att du ser vilka som väger mer eller mindre än medel.
3. **`.fit_transform()` = Gör båda två samtidigt.** Detta körs enbart på **träningsdatan**.

**Varför kör vi BARA `.transform()` på validerings- och testdatan?**  
När det kommer nya elever från en annan skola (testdata) vill du jämföra dem mot **din klass medelvärde (60 kg)**. Du får inte räkna ut ett nytt medelvärde för de nya eleverna, utan du måste använda den regelbok du redan skapat från träningsdatan.
