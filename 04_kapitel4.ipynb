**1. Vad kännetecknar klassificeringsproblem? Ge några exempel på tillämpningsområden**

Klassificering handlar om att lära en dator att dela in saker i bestämda grupper eller kategorier istället för att räkna ut ett tal.

Exempel:

E-post: Sortera mejl som antingen "Skräppost" eller "Viktigt".

Sjukvård: Avgöra om ett provsvar visar "Frisk" eller "Sjuk".

Bilder: Känna igen om en bild föreställer en "Hund", "Katt" eller "Bil".

**2. Förklara hur OvR- och OvO-algoritmerna fungerar.**

Många enkla datormodeller kan bara välja mellan två alternativ (ja/nej). För att de ska kunna välja mellan flera kategorier (t.ex. Hund, Katt, Häst) använder man två olika knep:

OvR (One-vs-Rest):
Datorn skapar en modell för varje djur.

Modell 1 testar: "Är det en Hund eller NÅGOT ANNAT?"

Modell 2 testar: "Är det en Katt eller NÅGOT ANNAT?"

Modell 3 testar: "Är det en Häst eller NÅGOT ANNAT?"
Den modell som känns säkrast på sin sak vinner.

OvO (One-vs-One):
Datorn ställer alla kategorier mot varandra två och två i en turnering.

Match 1: Hund vs Katt

Match 2: Hund vs Häst

Match 3: Katt vs Häst
Kategorin som vinner flest matcher vinner totalt.

**3. Förklara följande utvärderingsmått:**

a) Confusion matrix (Förväxlingsmatris):
En enkel tabell som visar hela sanningen. Den visar hur många gånger datorn gissade rätt och exakt vilka saker den förväxlade med varandra.

b) Accuracy (Total pricksäkerhet):
Hur stor andel av ALLA gissningar som var rätt.

Exempel: Om datorn gissar rätt på 90 av 100 mejl är pricksäkerheten 90 procent.

c) Precision (Träffsäkerhet):
När datorn väl säger "Det här är skräppost", hur ofta stämmer det?

Högt värde betyder: Datorn ropar sällan "varg" i onödan.

d) Recall (Fångstförmåga):
Hur stor andel av ALL den riktiga skräpposten lyckades datorn faktiskt hitta?

Högt värde betyder: Datorn missar nästan inga skräpmejl, men kan råka ta med några vanliga mejl av misstag.

e) F1-score (Balansmåttet):
Ett samlat betyg som slår ihop Precision och Recall till ett enda tal. Det visar hur väl datorn balanserar båda kraven.

f) ROC-kurvan:
Ett diagram som visar hur bra datorn är på att göra rätt jämfört med hur ofta den gör fel, beroende på hur sträng eller förlåtande man gör den.

**4. Vad är precision-recall tradeoff för något?**

Det handlar om en inbyggd kompromiss (en dragkamp). Du kan nästan aldrig få 100 procent på både Precision och Recall samtidigt.

Om du gör datorn väldigt sträng (Hög Precision):
Den stämplar bara ett mejl som skräppost om den är helt säker. Du slipper få viktiga mejl kastade i skräpkorgen, men en del skräppost kommer att slinka igenom till din inkorg (Låg Recall).

Om du gör datorn väldigt snäll (Hög Recall):
Den fångar upp all skräppost som finns, men den blir för ivrig och börjar kasta bort dina vanliga, viktiga mejl också (Låg Precision).

Du måste därför välja vad som är viktigast för just ditt problem.

**5. Några vanligt förekommande modeller för klassificeringsproblem är enligt
nedan. Förklara översiktligt hur respektive modell fungerar. Läs även ige-
nom respektive modells dokumentation, notera att du inte behöver förstå
alla detaljer från dokumentationen men det är bra att ha läst igenom den**

**a) Logistisk regression (LogisticRegression)**
Hur den fungerar: Trots namnet är detta en klassificeringsmodell. Den beräknar en linjär kombination av indata (features) och kör resultatet genom en s-formad funktion (sigmoid-funktionen) för att omvandla värdet till en sannolikhet mellan 0 och 1. Om sannolikheten överstiger en visströskel (oftast 0,5) klassificeras datapunkten som klass 1, annars klass 0.

**b) Support Vector Machines (SVC)**
Hur den fungerar: SVM försöker hitta en linje eller ett plan (hyperplan) som separerar klasserna med så stor marginal som möjligt till de närmaste datapunkterna (som kallas support vectors). Med hjälp av "kernel-tricket" kan modellen även hantera svåra, icke-linjära gränser genom att lyfta datan till högre dimensioner.

**c) Beslutsträd (DecisionTreeClassifier)**
Hur den fungerar: Skapar en trädliknande struktur av ja/nej-frågor baserat på indata (t.ex. "Är carat > 0.5?"). För varje fråga delas datan upp i allt renare delmängder tills den når ett slutgiltigt löv som ger klassen. Det är en mycket lättförståelig och visuell modell.

**d) Ensemble learning (VotingClassifier & BaggingClassifier)**
Hur det fungerar: Ensemble learning kombinerar flera enskilda modeller för att få en bättre och mer stabil förutsägelse än vad enstaka modeller klarar på egen hand.

VotingClassifier: Tränar helt olika modeller (t.ex. Logistisk regression, SVM och Beslutsträd) och låter dem rösta om den slutgiltiga klassen (majoritetsröstning).

BaggingClassifier: Tränar samma typ av modell flera gånger på slumpmässiga delmängder av träningsdatan (med återläggning) och tar genomsnittet eller röstningen av dem.

**e) Random Forest (RandomForestClassifier)**
Hur den fungerar: En vidareutveckling av Bagging med beslutsträd. Den bygger en hel "skog" av många oberoende beslutsträd. Varje träd tränas på ett slumpmässigt urval av datan och får bara välja bland ett slumpmässigt urval av attribut vid varje split. Slutgiltigt resultat blir majoritetsrösten från alla träd, vilket kraftigt minskar risken för överanpassning.

**f) Extra Trees (ExtraTreesClassifier)**
Hur den fungerar: Står för Extremely Randomized Trees. Fungerar väldigt likt Random Forest, men tar slumpmässigheten ett steg längre: istället för att leta efter den optimala tröskeln för att dela upp datan vid varje nod, väljer den uppdelningströsklar helt slumpmässigt. Detta gör träningen extremt snabb och kan ge ännu lägre varians.

Det handlar helt enkelt om att trädet hjälper oss att se vilka saker som är viktigast för slutresultatet.

**7. Stina säger till Kalle på lunchsamtalet “jag vill ha högsta möjliga precision
för vår klassificieringsmodell”. Kalle funderar ett tag och säger “men vad
händer då med recall”? Vad hade du svarat? I vilka fall kan man tänka sig
vilja ha en så hög precision som möjligt? I vilka fall kan det vara dåligt?
Om vi tänker oss rättsväsendet där en slutgiltig dom kan leda till fängelse,
vad kan vi då säga om precision-recall tradeoff?**

* **Svar till Kalle & Stina:**  
  Om vi tvingar modellen att ha så **hög precision som möjligt** blir den extremt försiktig och gissar bara när den är 100 % säker. Det gör att **recall sjunker** drastiskt – modellen blir så feg att den missar massor av riktiga fall.

* **När vill vi ha så hög precision som möjligt?**  
  När kostnaden för ett **falskt larm (False Positive)** är väldigt hög.  
  * *Exempel:* Ett skräppostfilter för e-post. Du vill inte att vanliga, viktiga mejl råkar hamnar i skräpkorgen.

* **När kan hög precision vara dåligt?**  
  När det är farligt eller dyrt att **missa ett riktigt fall (False Negative)**.  
  * *Exempel:* Cancerdiagnostik eller brandlarm. Det är bättre med ett falskt larm för mycket än att missa en sjuk patient eller en riktig brand.

* **Tillämpning i rättsväsendet (fängelsestraff):**  
  Här tillämpar man principen *"hellre fria än fälla"*. Det kräver **extremt hög precision** – vi vill inte sätta en oskyldig person i fängelse. Baksidan blir lägre recall: en del faktiska brottslingar kommer att gå fria för att bevisen inte räckte hela vägen upp till den höga kraven.

   * Eftersom klasserna överlappar varandra något i mitten syns en del gula punkter i det lila området och lila punkter i det gula området. Dessa representerar observationer där modellen har gissat fel på grund av att datan inte är helt perfekt separerbar med en rak linje.

### 9: Logiken bakom `.fit_transform()` vs `.transform()`

Principen handlar om att förhindra **dataläckage (data leakage)** så att modellen inte råkar "tjuvtitta" på testdatan i förväg.

Tänk dig att du ska räkna ut medelvikten på en skolklass:

1. **`.fit()` = Att lära sig reglerna.** Du väger alla i träningsklassen och räknar ut att medelvikten är **60 kg**.
2. **`.transform()` = Att använda reglerna.** Du drar av 60 kg från varje elevs vikt så att du ser vilka som väger mer eller mindre än medel.
3. **`.fit_transform()` = Gör båda två samtidigt.** Detta körs enbart på **träningsdatan**.

**Varför kör vi BARA `.transform()` på validerings- och testdatan?**  
När det kommer nya elever från en annan skola (testdata) vill du jämföra dem mot **din klass medelvärde (60 kg)**. Du får inte räkna ut ett nytt medelvärde för de nya eleverna, utan du måste använda den regelbok du redan skapat från träningsdatan.

### 13. I föregående kapitel gjordes en EDA på datasetet “hr_employee_data.xlsx”.
Gör nu ett komplett ML-flöde där den beroende variabeln, 𝑦, är left.
Om den variabeln är 1 så betyder det att den anställde har lämnat

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, RocCurveDisplay

# Sätt stil för visualiseringar
sns.set_theme(style="whitegrid")
%matplotlib inline
````

````python
# Ladda datasetet
df = pd.read_excel('datasets/hr_employee_data.xlsx')

# Visa form och första rader
print("Datasetets form:", df.shape)
print("\nFördelning av målvariabeln (left):")
print(df['left'].value_counts(normalize=True))

df.head()
````

````python
# Separera attribut (X) och målvariabel (y)
X = df.drop(columns=['left'])
y = df['left']

# Identifiera numeriska och kategoriska kolumner automatiskt
categorical_cols = X.select_dtypes(include=['object', 'category']).columns.tolist()
numerical_cols = X.select_dtypes(include=['int64', 'float64']).columns.tolist()

print("Numeriska variabler:", numerical_cols)
print("Kategoriska variabler:", categorical_cols)

# Skapa ColumnTransformer för skalning och One-Hot Encoding
preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), numerical_cols),
        ('cat', OneHotEncoder(drop='first', handle_unknown='ignore'), categorical_cols)
    ]
)

# Dela upp i tränings- och testdata (stratifierat eftersom y kan vara obalanserat)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print(f"\nTräningsdata: {X_train.shape[0]} rader")
print(f"Testdata:    {X_test.shape[0]} rader")
````

````python
# Skapa pipeline för Random Forest
pipeline_rf = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier(n_estimators=100, random_state=42, n_jobs=-1))
])

# Skapa pipeline för Logistisk Regression (basmodell)
pipeline_lr = Pipeline([
    ('preprocessor', preprocessor),
    ('classifier', LogisticRegression(random_state=42, max_iter=1000))
])

# Träna båda modellerna på träningsdatan
pipeline_rf.fit(X_train, y_train)
pipeline_lr.fit(X_train, y_train)

print("Modeller tränade framgångsrikt!")
````

````python
# Förutsäg på testdatan
y_pred_rf = pipeline_rf.predict(X_test)
y_pred_lr = pipeline_lr.predict(X_test)

print("=== Classification Report: Random Forest ===")
print(classification_report(y_test, y_pred_rf))

print("\n=== ROC-AUC Scores ===")
print("Random Forest ROC-AUC:", roc_auc_score(y_test, pipeline_rf.predict_proba(X_test)[:, 1]))
print("Logistic Regression ROC-AUC:", roc_auc_score(y_test, pipeline_lr.predict_proba(X_test)[:, 1]))
````

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Confusion Matrix för Random Forest
cm = confusion_matrix(y_test, y_pred_rf)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[0],
            xticklabels=['Jobbar kvar (0)', 'Har lämnat (1)'],
            yticklabels=['Jobbar kvar (0)', 'Har lämnat (1)'])
axes[0].set_title('Confusion Matrix - Random Forest')
axes[0].set_xlabel('Förutsagt')
axes[0].set_ylabel('Faktiskt')

# ROC-kurva
RocCurveDisplay.from_estimator(pipeline_rf, X_test, y_test, ax=axes[1], name='Random Forest')
RocCurveDisplay.from_estimator(pipeline_lr, X_test, y_test, ax=axes[1], name='Logistisk Regression')
axes[1].set_title('ROC-kurva Jämförelse')

plt.tight_layout()
plt.show()
````

````python
# Hämta feature importance från Random Forest
rf_model = pipeline_rf.named_steps['classifier']

# Skapa namn för alla variabler efter One-Hot Encoding
ohe_cols = pipeline_rf.named_steps['preprocessor'].named_transformers_['cat'].get_feature_names_out(categorical_cols)
feature_names = numerical_cols + list(ohe_cols)

# Skapa DataFrame för sortering
importances = pd.Series(rf_model.feature_importances_, index=feature_names).sort_values(ascending=False)

# Plotta de 10 viktigaste faktorerna
plt.figure(figsize=(10, 6))
importances.head(10).plot(kind='barh', color='teal')
plt.title('Top 10 Viktigaste Faktorer för Personalomsättning')
plt.xlabel('Vikt (Feature Importance)')
plt.gca().invert_yaxis()
plt.show()
````

### Preprocessing av egna foton

````python
from PIL import Image, ImageOps
from scipy import ndimage


def _to_grayscale(img):
    return img.convert("L")


def _auto_invert_if_needed(arr):
    """MNIST = vit siffra på svart botten. Foton är ofta tvärtom (mörk
    penna på ljust papper). Vi tittar på bildens kantpixlar (bakgrunden)
    och inverterar om de är ljusa."""
    border = np.concatenate([arr[0, :], arr[-1, :], arr[:, 0], arr[:, -1]])
    background_is_light = border.mean() > 127
    if background_is_light:
        arr = 255 - arr
    return arr


def _threshold(arr, margin=10):
    """Binärisera bilden med Otsu's metod (implementerad för hand,
    inga extra beroenden krävs)."""
    hist, _ = np.histogram(arr, bins=256, range=(0, 256))
    total = arr.size
    sum_total = np.dot(np.arange(256), hist)
    sum_b, w_b, max_var, threshold = 0.0, 0.0, 0.0, 0
    for t in range(256):
        w_b += hist[t]
        if w_b == 0:
            continue
        w_f = total - w_b
        if w_f == 0:
            break
        sum_b += t * hist[t]
        m_b = sum_b / w_b
        m_f = (sum_total - sum_b) / w_f
        var_between = w_b * w_f * (m_b - m_f) ** 2
        if var_between > max_var:
            max_var = var_between
            threshold = t
    return np.where(arr > max(threshold - margin, 0), arr, 0)
````

````python
def _crop_to_content(arr, pad=2):
    """Beskär bort tomma (svarta) marginaler runt siffran."""
    rows = np.any(arr > 0, axis=1)
    cols = np.any(arr > 0, axis=0)
    if not rows.any() or not cols.any():
        return arr
    rmin, rmax = np.where(rows)[0][[0, -1]]
    cmin, cmax = np.where(cols)[0][[0, -1]]
    rmin = max(rmin - pad, 0)
    cmin = max(cmin - pad, 0)
    rmax = min(rmax + pad, arr.shape[0] - 1)
    cmax = min(cmax + pad, arr.shape[1] - 1)
    return arr[rmin : rmax + 1, cmin : cmax + 1]


def _resize_keep_aspect(img, target=20):
    """MNIST-konvention: siffran fyller en 20x20-ruta innan den
    centreras i den slutliga 28x28-bilden."""
    w, h = img.size
    if w > h:
        new_w = target
        new_h = max(1, round(h * (target / w)))
    else:
        new_h = target
        new_w = max(1, round(w * (target / h)))
    return img.resize((new_w, new_h), Image.LANCZOS)


def _center_on_canvas(arr, canvas_size=28):
    """Placera siffran på en 28x28-canvas, centrerad utifrån dess
    tyngdpunkt (center of mass) - precis som MNIST gör."""
    h, w = arr.shape
    canvas = np.zeros((canvas_size, canvas_size), dtype=np.float64)
    top = (canvas_size - h) // 2
    left = (canvas_size - w) // 2
    canvas[top : top + h, left : left + w] = arr

    cy, cx = ndimage.center_of_mass(canvas)
    if not (np.isnan(cy) or np.isnan(cx)):
        shift_y = int(round(canvas_size / 2 - cy))
        shift_x = int(round(canvas_size / 2 - cx))
        canvas = ndimage.shift(canvas, shift=(shift_y, shift_x), cval=0)
    return canvas
````

````python
def preprocess_image(image, debug=False):
    """Ta emot en PIL-bild (t.ex. inläst fr책n ett mobilfoto) och
    returnera:
        - flat: (1, 784) float-vektor, redo för scaler.transform()
        - canvas: (28, 28) numpy-array, för visualisering
    """
    steps = {}

    gray = _to_grayscale(image)
    gray = ImageOps.exif_transpose(gray)  # fixa ev. EXIF-rotation från mobilkameror
    arr = np.array(gray).astype(np.float64)
    steps["gray"] = arr.copy()

    arr = _auto_invert_if_needed(arr)
    steps["inverted"] = arr.copy()

    arr = _threshold(arr)
    steps["thresholded"] = arr.copy()

    arr = _crop_to_content(arr)
    steps["cropped"] = arr.copy()

    cropped_img = Image.fromarray(arr.astype(np.uint8))
    resized_img = _resize_keep_aspect(cropped_img, target=20)
    resized_arr = np.array(resized_img).astype(np.float64)
    steps["resized"] = resized_arr.copy()

    canvas = _center_on_canvas(resized_arr, canvas_size=28)
    steps["final"] = canvas.copy()

    flat = canvas.reshape(1, -1)

    if debug:
        return flat, canvas, steps
    return flat, canvas
````
