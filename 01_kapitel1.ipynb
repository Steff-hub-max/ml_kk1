## Kapitel 1

**fråga 1: Hur hänger AI, ML och DL ihop**

AI är det övergripande området som utför uppgifter som verkar intelligenta.

ML är en metod inom AI som gör att datorn kan lära sig från data.

DL är en typ av ML som använder neurala nätverk för att lära sig mer komplexa mönster.

Ett exempel är att AI får en bild på ett djur. Med hjälp av ML kan datorn lära sig att känna igen att bilden föreställer en katt. Med DL kan den, genom att lära sig mer komplexa mönster i bilder, även exempelvis skilja mellan olika kattraser.

**fråga 3: förklara följande:**

**a) Var är syftet med att dela upp data i träningsdata, valideringsdata och testdata?**

Att kunna träna, justera och kontrollera hur bra en ML-modell fungerar på data den aldrig sett tidigare.

**b) Vad är k-delad korsvalidering för något?**

K-delad korsvalidering (k-fold cross-validation) är en metod för att kontrollera hur bra en ML-modell presterar och hur väl den kan generalisera till ny data.

**c) Vad är root mean squared error (RMSE) för något?**

Det är ett mått på hur stora fel en modell gör när den förutsäger numeriska värden.

**d) Vad är en hyperparameter för något? Vad är en parameter för något?**

En hyperparameter är ett värde som vi bestämmer innan eller under träningen för att styra hur modellen ska tränas.
En parameter är ett värde som modellen själv lär sig från träningsdatan.

**e) Vad är grid search och hur hänger namnet “grid” och “search” ihop med
processen som genomförs? Läser vi dokumentationen: https://scikit-learn.
org/stable/modules/generated/sklearn.model_selection.GridSearchCV så
ser vi att hyperparametern refit har standardvärdet True, vad innebär
det?**

Grid Search är en metod för att systematiskt testa olika kombinationer av hyperparametrar för att hitta den kombination som ger bäst resultat. Namnet "grid" kommer av att kombinationerna bildar ett rutnät, och "search" eftersom man söker igenom detta rutnät. refit=True innebär att när de bästa hyperparametrarna har hittats tränas modellen om på hela träningsdatan med dessa parametrar.

**f) Vad är kategorisk data och hur hanteras det? I ditt svar, använd begreppen
nominal data, ordinal data, one-hot-encoding, dummy-variable-encoding
och ordinal encoding.**

Kategorisk data måste ofta kodas om till numerisk data innan den kan användas av en ML-modell. För nominal data används ofta one-hot-encoding eller dummy-variable-encoding, medan ordinal data kan hanteras med ordinal encoding, eftersom ordningen mellan kategorierna har betydelse.

**g) Vad är feature engineering för något?**

Feature engineering är processen att bearbeta och skapa variabler så att datan blir mer användbar för en ML-modell och modellen kan göra bättre förutsägelser.

**h) Vad menas med principle of parsimony?**

Principle of parsimony betyder ungefär principen om enkelhet. Den innebär att man bör föredra den enklaste modellen som kan förklara datan tillräckligt bra.

**4. Vad menas med att “en modell är en förenkling av verkligheten”?**

En modell är en förenkling av verkligheten eftersom den bara använder ett begränsat antal variabler och antaganden för att beskriva eller förutsäga något i den verkliga världen. Den kan därför aldrig fånga alla faktorer som påverkar verkligheten.

**5. Vad menas med att en modell är “överanpassad” eller overfitted på engels-
ka?**

En modell är överanpassad (overfitted) när den har lärt sig träningsdatan för bra, inklusive detaljer och slumpmässiga variationer som egentligen inte är användbara.

Det gör att modellen kan prestera väldigt bra på träningsdatan, men betydligt sämre på ny data som den inte har sett tidigare.

**6. Högre är bättre i scikit-learn scoring, vad innebär det?**

det betyder att scikit-learns scoring-funktioner alltid ska ge ett högre värde för en bättre modell. För mått där lägre egentligen är bättre, som RMSE eller MAE, använder scikit-learn negativa värden.

**7. Vad är tvärsnittsdata, tidsseriedata och paneldata? Exemplifiera när re-
spektive datakategori kan uppstå.**

Tvärsnittsdata är data från många olika individer eller objekt vid ungefär samma tidpunkt.
tidsseriedata är data från samma objekt eller variabel uppmätt vid flera tidpunkter.
paneldata är en kombination av tvärsnittsdata och tidsseriedata, så man följer flera individer eller objekt över flera tidpunkter.

**8. Ge några exempel på verkliga tillämpningsområden inom ML. Sök gärna
på nätet för att besvara frågan.**

Machine Learning används inom många områden, exempelvis rekommendationssystem som Netflix och Spotify, bedrägeribekämpning inom bankväsendet, sjukvård för analys av medicinska bilder och förutsägelse av sjukdomsrisker, spamfilter, självkörande fordon, ansikts- och bildigenkänning samt röstassistenter och cybersäkerhet.

**9. Generellt sett gäller det att högre är bättre i scikit-learn scoring, därför
används exempelvis scoring='neg_mean_squared_error'. Förklara lo-
giken bakom detta, det vill säga att vi använder “negative” mean squared
error.**

MSE ska egentligen minimeras, men scikit-learn maximerar sina scoring-värden. Därför används det negativa värdet neg_mean_squared_error, så att en modell med lägre MSE får ett högre (mindre negativt) scoring-värde.

**11. I koden nedan använder vi först test_size=0.2 och sedan test_size=0.25,
förklara varför det ger oss proportionerna 60-20-20**

test_size=0.2 ger först 80 % träning och 20 % test. Därefter delar test_size=0.25 upp de 80 % i 25 % validering och 75 % träning. 0.25 × 80 = 20, vilket ger 60-20-20.
